```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
document =["Eat healthy food and exercise daily for a healthy lifestyle. Some healthy food are vegetables lentils etc"]
```

```python
bow = CountVectorizer()
bow_matrix = bow.fit_transform(document)
bow_matrix
```

```python
print(bow_matrix.toarray())
```

```python
bow.get_feature_names_out()
```

```python
documents = ['people watch movie and watch movie again',
 'people watch cricket and watch cricket',
 'people like movie and like movie a lot',
 'people like cricket']
```

```python
bow_matrix = bow.fit_transform(documents)
```

```python
bow.get_feature_names_out()
```
