```python
from sklearn.preprocessing import OneHotEncoder
import numpy
```

```python
document = ["my name is vishal I love ai"]

documents = [
    "people watch movie",
    "people watch cricket",
    "people love movie",
    "people love cricket"
]
```

```python
tokens = [sentence.lower().split() for sentence in documents]
tokens
```

```python
all_words = [[word] for sentence in tokens for word in sentence]
all_words
```

```python
encoder = OneHotEncoder(sparse_output=False)
```

```python
encoder.fit(all_words)
encoder.categories_
```

```python
print("Vocabulary:",encoder.categories_[0])
```

```python
for sentence in tokens:
    encoded_sentence = encoder.transform([[word] for word in sentence])   
    print("\nSentence:",sentence)
    print(encoded_sentence)
```

```python
print(encoder.transform([["love"],["watch"]]))
```
