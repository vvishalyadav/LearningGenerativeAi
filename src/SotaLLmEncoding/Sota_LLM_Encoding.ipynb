ViT(vision transformer) and CLIP(contrastive language-image pretraining) are two popular transformer-based models used for image processing tasks.

```python
from transformers import CLIPModel, CLIPProcessor
```

```python
from PIL import Image
```

```python
image = Image.open(r'D:\Vishal Projs\My_Projects\SeptPy\LearningGenerativeAi\public\images\MachineImg.png')
```

```python
processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")
```

```python
model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32")
```

```python
inputs = processor(images=image, return_tensors='pt')
```

```python
import torch
```

```python
with torch.no_grad():
    image_embedding = model.get_image_features(**inputs)
```
